### Dimensionsanalyse - ét dimensionsløst produkt

Lad os sige at vi er interesseret i at finde en model for $x=f(F,m,v)$ hvor $x$ er en forskydning, $F$ er en kraft, $m$ er en masse $v$ er en fart, $a$ er en acceleration og $t$ er en tid.
Vi kan omformere dette til en funktion $F(x,F,m,v)=0$.
Som det første bestemmer vi dimensionerne af alle de variable (både afhængig og uafhængige), 
hvor $\text{M}$ indikerer massedimension,
hvor $\text{L}$ indikerer længdedimension
og 
hvor $\text {T}$ indikerer tidsdimension.

I dette tilfælde har vi enhederne for de variable. Bemærk, at vi kun opererer med grunddimensionerne, dvs. vi ikke bruger newton, watt, joule etc. 
Disse udtrykkes ved grunddimensionerne som $\text {N}=\text {M}\text {L}\text {T}^{-2}$ og $\text {J}=\text {M} \text {L}^2\text {T}^{-2}$.
$$\left[x\right]=\text {L}$$
$$\left[F\right]=\text {M}\text {L}\text {T}^{-2}$$
$$\left[m\right]=\text {M}$$
$$\left[v\right]=\text {L}\text {T}^{-1}$$

Vi konstruerer nu den såkaldte dimensionsmatrix (her kaldet $M$, vi benytter normalt $A$ men her indgår et andet $A$)
$$M=\begin{bmatrix}0 &1 & 1 & 0\\1 & 1 & 0 & 1 \\0 &-2 & 0&-1 \end{bmatrix}$$
Første række indeholder eksponenterne fra massedimensionenerne af de variable og
anden række indeholder eksponenterne fra længdedimensionenerne af de variable og
de, tredje række indeholder eksponenter fra tidsdimensionenrne af de variable.
Rækkefølgen af de variable (søjler) er at man skriver den afhængige variabel først og dernæste de uafhængige variable,
vi har her valgt rækkefølgen $x,F,m,v$.

Centralt i løsningen af problemet er de såkaldte dimensionsløse parametre der er af formen
$$\pi=x^AF^Bm^Cv^D$$
hvor eksponenterne $A,B,C,D$ skal bestemmes så udtrykket bliver dimensionsløst, dvs. $\left[\pi\right]=1$.

Eksponenterne er løsninger til det homogene ligningssystem
$$M\begin{bmatrix}A\\B\\C\\D\end{bmatrix}=\begin{bmatrix}1 &1 & 1 & 0\\0 & -1 & -2 & 1 \end{bmatrix}\begin{bmatrix}A\\B\\C\\D\end{bmatrix}=\begin{bmatrix}A+B+C\\-B-2C+D \end{bmatrix}
=\begin{bmatrix}0\\0 \end{bmatrix}$$

Der er normalt uendelig mange løsninger men det er antallet af lineært uafhængige løsninger der er interessant.
Hvis der er $n$ variable (her har vi $n=4$) så er antallet af lineært uafhængige løsninger
$n-\text{rank(M)}$

Som ses nedenfor er rangen to så der skal findes to løsninger (der er lineært uafhængige).

```python
import sympy as sp
A,B,C,D = sp.symbols('A:D')
M = sp.Matrix([[0,1,1,0],[1,1,0,1],[0,-2,0,-1]])
print(M.rank())
```

```python
import sympy as sp
x,F,m,v = sp.symbols('x,F,m,v')
A,B,C,D = sp.symbols('A:D')
M = sp.Matrix([[0,1,1,0],[1,1,0,1],[0,-2,0,-1]])
display(M)
sol = sp.solve([M*sp.Matrix([A,B,C,D]),sp.Eq(A,1)],dict=True)
print(sol)
display(x**sol[0][A]*F**sol[0][B]*m**sol[0][C]*v**sol[0][D])
```

```python
pi = x**sol[0][A] * F**sol[0][B] * m**sol[0][C] * v**sol[0][D]
display(sp.Eq(sp.symbols('pi'),pi))
k = sp.symbols('k')
eq = sp.Eq(pi,k)
display(eq)
sol = sp.solve(eq,x,dict=True)
display(sp.Eq(x,sol[0][x]))
```

### Dimensionsanalyse - to dimensionsløse produkter

Lad os sige at vi er interesseret i at finde en model for $x=f(v,a,t)$ hvor $x$ er en forskydning, $v$ er en hastighed, $a$ er en acceleration og $t$ er en tid.
Vi kan omformere dette til en funktion $F(x,v,a,t)=0$.
Som det første bestemmer vi dimensionerne af alle de variable (både afhængig og uafhængige), 
hvor $\text M$ indikerer massedimension,
hvor $\text L$ indikerer længdedimension
og 
hvor $\text T$ indikerer tidsdimension.

I dette tilfælde har vi enhederne for de variable. Bemærk, at vi kun opererer med grunddimensionerne, dvs. vi ikke bruger newton, watt, joule etc. 
Disse udtrykkes ved grunddimensionerne som $\text N=\text M \text L\text T^{-2}$ og $\text J=\text M \text L^2\text T^{-2}$.
$$\left[x\right]=\text L$$
$$\left[v\right]=\text L\text T^{-1}$$
$$\left[a\right]=\text L\text T^{-2}$$
$$\left[t\right]=\text T$$

Vi konstruerer nu den såkaldte dimensionsmatrix (her kaldet $M$, vi benytter normalt $A$ men her indgår et andet $A$)
$$M=\begin{bmatrix}1 &1 & 1 & 0\\0 & -1 & -2 & 1 \end{bmatrix}$$
Første række indeholder eksponenterne fra længdedimensionen af de variable og
de anden række indeholder eksponenter fra tidsdimensionen af de variable.
Rækkefølgen af de variable (søjler) er at man skriver den afhængige variabel først og dernæste de uafhængige variable,
vi har her valgt rækkefølgen $x,v,a,t$.

Centralt i løsningen af problemet er de såkaldte dimensionsløse parametre der er af formen
$$\pi=x^Av^Ba^Ct^D$$
hvor eksponenterne $A,B,C,D$ skal bestemmes så udtrykket bliver dimensionsløst, dvs. $\left[\pi\right]=1$.

Eksponenterne er løsninger til det homogene ligningssystem
$$M\begin{bmatrix}A\\B\\C\\D\end{bmatrix}=\begin{bmatrix}1 &1 & 1 & 0\\0 & -1 & -2 & 1 \end{bmatrix}\begin{bmatrix}A\\B\\C\\D\end{bmatrix}=\begin{bmatrix}A+B+C\\-B-2C+D \end{bmatrix}
=\begin{bmatrix}0\\0 \end{bmatrix}$$

Der er normalt uendelig mange løsninger men det er antallet af lineært uafhængige løsninger der er interessant.
Hvis der er $n$ variable (her har vi $n=4$) så er antallet af lineært uafhængige løsninger
$n-\text{rank(M)}$

Som ses nedenfor er rangen to så der skal findes to løsninger (der er lineært uafhængige).

```python
import sympy as sp
A,B,C,D = sp.symbols('A:D')
M = sp.Matrix([[1,1,1,0],[0,-1,-2,1]])
print(M.rank())
```

Da der er to ligninger med fire ubekendte kan vi specificere to værdier.
For at sikre sig at de to dimensionsløse parametre vi bestemmer er uafhængige 
så vælger vi at specificere at $A=1$ for den ene dimensionsløse parameter og
at $A=0$ for den anden dimensionsløse parameter.
Dermed sikrer vi os at de to løsninger er lineært uafhængige.
Vi skal dog stadig vælge en ekstra af eksponenterne. 
Her vælger vi at sætter $B=0$ sammen med $A=1$ og $B=1$ sammen med $A=0$.

```python
import sympy as sp
x,v,a,t = sp.symbols('x,v,a,t')
A,B,C,D = sp.symbols('A:D')
M = sp.Matrix([[1,1,1,0],[0,-1,-2,1]])
display(M)
p1 = sp.solve([M*sp.Matrix([A,B,C,D]),sp.Eq(A,1),sp.Eq(B,0)],dict=True)
p2 = sp.solve([M*sp.Matrix([A,B,C,D]),sp.Eq(A,0),sp.Eq(B,1)],dict=True)
print(p1,p2)
display(x**p1[0][A]*v**p1[0][B]*a**p1[0][C]*t**p1[0][D])
display(x**p2[0][A]*v**p2[0][B]*a**p2[0][C]*t**p2[0][D])
```

Løsningen på problemet er så
$$F\left(\frac{x}{at^2},\frac{v}{at}\right)=0$$

Vi kan isolere den afhængige varibel, $x$, og finder
$$x=at^2 H\left(\frac{v}{at}\right)$$


Vi prøver nu at finde en anden løsning.
Her vælger vi at sætte $C=0$ og $C=1$ i stedet for det samme med $B$.

```python
import sympy as sp
x,v,a,t = sp.symbols('x,v,a,t')
A,B,C,D = sp.symbols('A:D')
M = sp.Matrix([[1,1,1,0],[0,-1,-2,1]])
display(M)
p1 = sp.solve([M*sp.Matrix([A,B,C,D]),sp.Eq(A,1),sp.Eq(C,0)],dict=True)
p2 = sp.solve([M*sp.Matrix([A,B,C,D]),sp.Eq(A,0),sp.Eq(C,1)],dict=True)
print(p1,p2)
display(x**p1[0][A]*v**p1[0][B]*a**p1[0][C]*t**p1[0][D])
display(x**p2[0][A]*v**p2[0][B]*a**p2[0][C]*t**p2[0][D])
```

Her får vi løsningen
$$F\left(\frac{x}{vt},\frac{at}{v}\right)=0$$

Vi kan isolere den afhængige varibel og finder
$$x=vt H\left(\frac{at}{v}\right)$$

Vi kan også finde en generel løsning hvor løsninger er parametriseret med to parametre.

```python
import sympy as sp
x,v,a,t = sp.symbols('x,v,a,t')
A,B,C,D = sp.symbols('A:D')
M = sp.Matrix([[1,1,1,0],[0,-1,-2,1]])
display(M)
p1 = sp.solve([M*sp.Matrix([A,B,C,D])],dict=True)
p2 = sp.solve([M*sp.Matrix([A,B,C,D])],dict=True)
print(p1,p2)
display(x**p1[0][A]*v**p1[0][B]*a**C*t**D)
display(x**p2[0][A]*v**p2[0][B]*a**C*t**D)
```

Her en list mere generel måde at skrive det op på, læs det kun hvis du er god til numpy.

```python
import numpy as np
import sympy as sp
x,v,a,t = sp.symbols('x,v,a,t')
A,B,C,D = sp.symbols('A:D')
M = sp.Matrix([[1,1,1,0],[0,-1,-2,1]])
display(M)
p1 = sp.solve([M*sp.Matrix([A,B,C,D]),sp.Eq(A,1),sp.Eq(B,0)],dict=True)
p2 = sp.solve([M*sp.Matrix([A,B,C,D]),sp.Eq(A,0),sp.Eq(B,1)],dict=True)
ps = [p1,p2] # list of solutions
vars = [x,v,a,t] # list of variables
pis = []
for p in ps:
    pis.append(np.prod(np.array(vars)**np.array(list(p[0].values()))))
print(pis)
for pi in pis:
    display(pi)
```
