```python
import sympy as sp

A,B,C,D = sp.symbols('A,B,C,D')
pi = sp.symbols('pi')
```

```python
# 1a

v,h,g = sp.symbols('v,h,g')
k = sp.symbols('k')

M = sp.Matrix([[1,1,1],[-1,0,-2]])
print('Dimensionsløse parametre:', 3-M.rank())
sol = sp.solve([M * sp.Matrix([A,B,C]), sp.Eq(A,1)],dict=True)
print(sol)

pi1 = v**sol[0][A]*h**sol[0][B]*g**sol[0][C]
display(sp.Eq(pi,pi1))
eq = sp.Eq(pi1,k)
display(eq)
sol = sp.solve(eq,v,dict=True)
display(sp.Eq(v,sol[0][v]))
```

```python
#2b

F,p,v = sp.symbols('F,p,v')
M = sp.Matrix([[0,1,1],[1,1,-1],[-1,-2,0]])
display(M)
print('Dimensionsløse parametre:', 3-M.rank())
sol = sp.solve([M * sp.Matrix([A,B,C]), sp.Eq(A,1)],dict=True)
pi1 = v**sol[0][A]*F**sol[0][B]*p**sol[0][C]
display(sp.Eq(pi,pi1))
eq = sp.Eq(pi1,1)
display(eq)
sol = sp.solve(eq,v,dict=True)
display(sp.Eq(v,sol[0][v]))
```

```python
#3
L,c,G,h = sp.symbols('L,c,G,h')
M = sp.Matrix([[0,0,-1,1],[1,1,3,2],[0,-1,-2,-1]])
display(M)
print('Dimensionsløse parametre:', 4-M.rank())
sol = sp.solve([M* sp.Matrix([A,B,C,D]), sp.Eq(A,1)],dict=True)
pi1 = L**sol[0][A]*c**sol[0][B]*G**sol[0][C]*h**sol[0][D]
display(sp.Eq(pi,pi1))
eq = sp.Eq(pi1,1)
display(eq)
sol = sp.solve(eq,L,dict=True)
display(sp.Eq(L,sol[0][L]))
```
